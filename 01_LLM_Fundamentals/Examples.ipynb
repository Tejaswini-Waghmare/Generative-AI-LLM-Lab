## Token frequency baseline — not a neural tokenizer

```python
from collections import Counter
corpus='Models learn patterns from examples and examples reveal patterns'
tokens=corpus.lower().split()
print(Counter(tokens))
print('Naive whitespace token count:',len(tokens))
```

```python
rng=np.random.default_rng(7)
counts=Counter(tokens)
words=list(counts); weights=np.array(list(counts.values()),dtype=float); weights/=weights.sum()
print('Sampled baseline continuation:', ' '.join(rng.choice(words,5,p=weights)))
```
