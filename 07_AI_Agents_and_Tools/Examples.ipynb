## Safe deterministic tool router

```python
import re
def calculator(expression):
    # Deliberately narrow contract; no eval or arbitrary code execution
    m=re.fullmatch(r'\s*(\d+(?:\.\d+)?)\s*([+*])\s*(\d+(?:\.\d+)?)\s*',expression)
    if not m:return 'Unsupported expression'
    a,op,b=m.groups();return float(a)+float(b) if op=='+' else float(a)*float(b)
def route(request):
    if request.startswith('calc:'):return calculator(request[5:])
    return 'Escalate: no permitted tool for this request'
print(route('calc: 12 * 4'))
print(route('delete all files'))
```
